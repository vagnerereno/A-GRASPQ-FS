# A-GRASPQ-FS Demo

This notebook demonstrates how to use A-GRASPQ-FS as a plug-and-play feature selector compatible with scikit-learn.

The examples below show that the algorithm can select different subset sizes and feature combinations depending on the dataset.

### Imports

```python
from agraspqfs import AGraspQFeatureSelector

from sklearn.datasets import load_breast_cancer, load_wine
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import f1_score
```

## Dataset 1 — Breast Cancer
In this first example, we use the Breast Cancer dataset, a binary classification problem.
A-GRASPQ-FS searches for both the number of selected features and the feature subset itself.

```python
X, y = load_breast_cancer(return_X_y=True, as_frame=True)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    stratify=y,
    random_state=42,
)

selector = AGraspQFeatureSelector(
    estimator=GaussianNB(),
    preset="fast",
    min_features=2,
    max_features=15,
    random_state=42,
)

selector.fit(X_train, y_train)

clf = GaussianNB().fit(selector.transform(X_train), y_train)
pred = clf.predict(selector.transform(X_test))
holdout_f1 = f1_score(y_test, pred)

print("Dataset: Breast Cancer")
print("Original number of features:", X_train.shape[1])
print("Selected feature names:", selector.selected_features_)
print("Number of selected features:", selector.n_features_selected_)
print("Best internal score:", round(selector.best_score_, 4))
print("Holdout F1-score:", round(holdout_f1, 4))
```

## Dataset 2 — Wine

In this second example, we use the Wine dataset, a multiclass classification problem.
Since the dataset is different, A-GRASPQ-FS may converge to a different subset size and a different set of selected features.

```python
X, y = load_wine(return_X_y=True, as_frame=True)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    stratify=y,
    random_state=42,
)

selector = AGraspQFeatureSelector(
    estimator=GaussianNB(),
    preset="fast",
    min_features=2,
    max_features=15,
    random_state=42,
)

selector.fit(X_train, y_train)

clf = GaussianNB().fit(selector.transform(X_train), y_train)
pred = clf.predict(selector.transform(X_test))
holdout_f1 = f1_score(y_test, pred, average="weighted")

print("Dataset: Wine")
print("Original number of features:", X_train.shape[1])
print("Selected feature names:", selector.selected_features_)
print("Number of selected features:", selector.n_features_selected_)
print("Best internal score:", round(selector.best_score_, 4))
print("Holdout weighted F1-score:", round(holdout_f1, 4))
```

## Takeaway

The selected subset size is not fixed. A-GRASPQ-FS jointly searches for the subset cardinality and the feature composition. Therefore, different datasets and classifiers may lead to different selected feature subsets.
